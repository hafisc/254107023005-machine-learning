Pada praktikum keempat di Jobsheet 3 ini saya melakukan ekstraksi fitur sederhana pada data tidak terstruktur berupa citra `Lenna.png` menggunakan library `Pillow` (`PIL`), yaitu memisahkan channel warna RGB dan mengambil nilai fitur histogram pada channel warna merah (*Red*).

### Langkah 1 - Load Image

Membuka file citra menggunakan `PIL.Image.open()` dan menampilkannya di notebook.

```python
from PIL import Image

img = Image.open('img/Lenna_(test_image).png')
display(img) # tampilkan gambar di notebook
```

### Langkah 2 - Ekstrak Fitur

Memisahkan citra menjadi tiga channel warna (`r`, `g`, `b`) menggunakan `img.split()`, lalu mengecek panjang ukuran serta mencetak nilai array histogram pada channel *Red* (`r.histogram()`).

```python
# Ekstrak setiap channel red, green, blue
r, g, b = img.split()

# Cek panjang ukuran channel red
print(len(r.histogram()))

# Cetak fitur histogram pada channel red
print(r.histogram())
```

Dari hasil ekstraksi di atas, fungsi `r.histogram()` menghasilkan sebuah array sepanjang `256` elemen (indeks intensitas piksel `0` sampai `255`). Setiap angka di dalam array tersebut merepresentasikan frekuensi kemunculan nilai intensitas merah tertentu pada citra. Terlihat bahwa untuk nilai intensitas merah yang sangat gelap (indeks `0` sampai sekitar `40`) jumlah pikselnya bernilai `0`, sedangkan frekuensi piksel paling banyak terkonsentrasi pada rentang intensitas menengah hingga tinggi (indeks `180` sampai `235`), yang sesuai dengan dominasi warna kemerahan/hangat pada citra Lenna.
