Pada praktikum kedua di Jobsheet 3 ini saya melakukan pra pengolahan data terlebih dahulu sebelum melakukan seleksi fitur pada dataset `Titanic-Dataset-fixed.csv`. Tahapan pra pengolahan meliputi encoding pada fitur `Sex` (mapping `'male'` menjadi `1` dan `'female'` menjadi `0`), penyederhanaan fitur `Cabin` menjadi huruf pertama kode dek, demonstrasi imputasi median pada `Age`, serta standardisasi pada `Age` menggunakan `StandardScaler`, kemudian diakhiri dengan pemilihan 5 fitur utama (`Survived`, `Pclass`, `Age`, `Sex`, `Cabin`).

### Langkah 0 - Load Library

Mengimpor library `pandas`, `LabelEncoder`, dan `StandardScaler`.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

### Langkah 1 - Load Data

Memuat dataset `Titanic-Dataset-fixed.csv` dan menampilkan 5 data teratas.

```python
dpath = 'Titanic-Dataset-fixed.csv'
df = pd.read_csv(dpath)
df.head()
```

### Langkah 2 & 3 - Pre Processing Encoding `Sex` dan Verifikasi

Melakukan pemetaan (*mapping*) nilai kategorikal pada kolom `Sex` menjadi numerik biner (`'male': 1`, `'female': 0`).

```python
# --- 2. Encoding fitur 'Sex' ---
df['Sex'] = df['Sex'].map({'male': 1, 'female': 0})
print("=== Hasil Setelah Encoding 'Sex' ===")
print(df[['Sex']].head(), "\n")
```

### Langkah 4 & 5 - Pre Processing Encoding `Cabin` dan Verifikasi

Mengisi nilai kosong (jika ada) pada `Cabin` dengan `'Unknown'`, kemudian mengekstrak karakter pertama dari string `Cabin` untuk merepresentasikan kode dek kapal.

```python
# --- 3. Encoding fitur 'Cabin' ---
df['Cabin'] = df['Cabin'].fillna('Unknown')
print("=== Hasil Setelah Fill NaN pada 'Cabin' ===")
print(df[['Cabin']].head(5), "\n")
df['Cabin'] = df['Cabin'].apply(lambda x: x[0] if x != 'Unknown' else 'U')
print("=== Hasil Ambil huruf pertama 'Cabin' (Deck) ===")
print(df[['Cabin']].head(5), "\n")
```

### Langkah 5 & 6 - Pre Processing Imputasi Data `Age` dan Verifikasi

Mendemonstrasikan proses imputasi nilai kosong pada kolom `Age` menggunakan nilai tengah (*median*).

```python
# --- 4. Imputasi fitur 'Age' dengan median ---
df['Age'] = df[['Age']].fillna(df['Age'].median())
print("=== Hasil Setelah Imputasi Data pada Usia dengan median ===")
print(df[['Age']].head(10), "\n")
```

### Langkah 7 & 6 - Pre Processing Standarisasi `Age` dan Verifikasi

Melakukan standardisasi nilai pada kolom `Age` menggunakan `StandardScaler`.

```python
scaler = StandardScaler()
df['Age'] = scaler.fit_transform(df[['Age']])
print("=== Hasil Setelah Standardisasi 'Age' ===")
print(df[['Age']].head(10), "\n")
```

### Langkah 7 & 8 - Hasil setelah Pre Processing

Mengecek tampilan 5 baris pertama keseluruhan DataFrame setelah dilakukan perubahan pada kolom `Sex`, `Cabin`, dan `Age`.

```python
# --- 5. Cek hasil akhir ---
print("=== Hasil Preprocessing (5 baris pertama) ===")
#print(df[['Sex', 'Cabin', 'Age']].head(5))
df.head()
```

### Langkah 9 - Seleksi Fitur

Memilih 5 variabel penting yang relevan terhadap keselamatan penumpang, yaitu `Survived`, `Pclass`, `Age`, `Sex`, dan `Cabin`.

```python
df = df[['Survived', 'Pclass', 'Age', 'Sex', 'Cabin']]
df.head()
```
