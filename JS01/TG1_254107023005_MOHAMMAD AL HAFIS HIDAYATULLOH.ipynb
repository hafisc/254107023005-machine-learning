### 1.1 PyPREP

```python
!pip install pyprep
```

PyPREP ini implementasi Python dari PREP pipeline yang khusus buat preprocessing data EEG. Jadi gini, kalau kerja dengan sinyal otak, data mentahnya itu biasanya kotor banget, banyak noise dari gerakan mata, kedipan, bahkan dari listrik di sekitar ruangan. Library ini bantu bersihin data EEG secara sistematis lewat beberapa tahap kayak identifikasi channel yang rusak, interpolasi channel, re-referencing yang robust, sama deteksi artifact. Yang menarik dari PyPREP adalah metodenya sudah terstandarisasi jadi hasil preprocessing kita bisa lebih konsisten dan reproducible, penting banget ini kalau mau publikasi penelitian atau kolaborasi dengan tim lain.

Dalam konteks machine learning, preprocessing yang baik itu fondasi penting. Model ML kita cuma bisa sebagus data yang kita kasih. PyPREP memastikan data EEG yang masuk ke model kita sudah bersih dan siap diproses, yang pada akhirnya akan meningkatkan performa model untuk aplikasi seperti brain-computer interface, deteksi epilepsi, atau analisis cognitive state.

### 1.2 SciPy

### 1.4 pyECG (py-ecg-detectors)

```python
!pip install py-ecg-detectors
```

pyECG adalah library yang berisi koleksi algoritma untuk deteksi heartbeat dari sinyal ECG. Sinyal ECG itu merekam aktivitas listrik jantung, dan salah satu tugas paling fundamental dalam analisis ECG adalah mendeteksi R-peaks yaitu puncak gelombang yang menandakan detak jantung.

Library ini mengimplementasikan 8 algoritma deteksi yang udah established di penelitian biomedical, misalnya Pan-Tompkins yang merupakan salah satu algoritma paling populer menggunakan filtering dan thresholding adaptif, Hamilton yang metodenya robust untuk sinyal dengan noise, Christov yang merupakan algoritma real-time yang efisien, dan algoritma lainnya seperti Engelse-Zeelenberg, SWT (Stationary Wavelet Transform), Two Average, Matched Filter, sama WQRS.

Selain deteksi R-peaks, pyECG juga punya modul HRV (Heart Rate Variability) analysis yang bisa menghitung berbagai metrics seperti RMSSD, SDNN, pNN50, dan frequency domain analysis. Ini berguna banget untuk aplikasi kesehatan seperti monitoring stress level, deteksi arrhythmia, atau bahkan prediksi risiko kardiovaskular. Dalam konteks machine learning, output dari pyECG kayak heart rate dan HRV metrics bisa dijadikan features untuk model klasifikasi atau prediksi kondisi kesehatan. Misalnya model untuk deteksi atrial fibrillation, prediksi sleep stages, atau monitoring pasien ICU. Jadi pyECG ini tools penting untuk feature extraction dari sinyal biomedis.

## Bagian 2: Etika dan Dampak Lingkungan AI

### 2.1 Pelanggaran Etika dan Hukum dalam Penggunaan Kecerdasan Buatan

Kalau kita lihat perkembangan AI dalam beberapa tahun terakhir memang banyak breakthrough yang impressive. Tapi di balik kemajuan itu ada beberapa kasus yang cukup mengkhawatirkan terkait bias, diskriminasi, dan pelanggaran privasi. Berikut beberapa contoh nyata yang terdokumentasi dengan baik.

Salah satu studi paling terkenal tentang bias AI adalah penelitian Gender Shades oleh Joy Buolamwini dan Timnit Gebru dari MIT Media Lab. Mereka mengevaluasi tiga sistem facial recognition komersial dari perusahaan teknologi besar yaitu IBM, Microsoft, dan Face++, dan menemukan disparitas akurasi yang signifikan. Hasilnya cukup mengejutkan karena algoritma-algoritma tersebut punya error rate yang jauh lebih tinggi untuk wanita berkulit gelap dibanding pria berkulit terang. Secara spesifik error rate untuk dark-skinned females mencapai 34.7% sedangkan untuk light-skinned males hanya 0.8%, artinya sistem ini 40 kali lebih sering salah mengidentifikasi wanita berkulit gelap dibanding pria berkulit terang.

Masalahnya bukan cuma soal akurasi teknis. Teknologi facial recognition ini dipakai untuk berbagai aplikasi kritis dari unlock smartphone sampai surveillance oleh law enforcement. Kalau sistemnya bias dampaknya bisa serius, misidentifikasi yang bisa berujung pada penangkapan orang yang salah. Buolamwini dan Gebru menunjukkan bahwa bias ini muncul karena dataset training yang tidak seimbang, mayoritas berisi wajah orang kulit putih dan laki-laki.

Contoh lain yang cukup controversial adalah COMPAS (Correctional Offender Management Profiling for Alternative Sanctions), sistem AI yang digunakan di pengadilan Amerika untuk memprediksi risiko seorang terdakwa melakukan kejahatan lagi atau recidivism. Investigasi oleh ProPublica pada 2016 menemukan bahwa COMPAS memiliki bias rasial yang signifikan. Analisis mereka terhadap lebih dari 7,000 kasus menunjukkan bahwa terdakwa kulit hitam dua kali lebih mungkin salah dilabeli sebagai high risk dibanding terdakwa kulit putih dengan criminal history yang serupa. Sebaliknya terdakwa kulit putih yang dilabeli low risk ternyata lebih sering melakukan kejahatan lagi dibanding terdakwa kulit hitam dengan skor risiko yang sama.

Yang bikin ini problematic adalah COMPAS digunakan oleh hakim untuk menentukan vonis termasuk keputusan tentang bail dan hukuman penjara. Bias dalam algoritma berarti ada orang yang mungkin mendapat hukuman lebih berat bukan karena tindakan mereka tapi karena warna kulit mereka. Ini jelas melanggar prinsip keadilan dan persamaan di depan hukum.

Kasus-kasus di atas menunjukkan beberapa masalah fundamental. Pertama, bias data sama dengan bias model. Algoritma ML belajar dari data historis yang mengandung bias sosial, kalau kita train model dengan data yang bias model akan mereplikasi dan bahkan memperkuat bias tersebut. Kedua, kurangnya transparansi karena banyak sistem AI yang proprietary dan ngga transparent. COMPAS misalnya adalah black box, bahkan hakim yang menggunakannya ngga tahu persis bagaimana algoritma itu bekerja. Ketiga, tanggung jawab hukum yang ambigu, kalau AI salah prediksi dan merugikan seseorang siapa yang bertanggung jawab, developer, perusahaan, atau user? Framework hukum kita masih belum jelas soal ini. Keempat, penggunaan AI yang bias di sektor-sektor kritis seperti law enforcement, hiring, healthcare bisa memperparah ketidakadilan yang sudah ada di masyarakat.

Beberapa yurisdiksi mulai merespons masalah ini. EU dengan AI Act-nya misalnya mengklasifikasikan sistem AI berdasarkan tingkat risiko dan menerapkan regulasi yang ketat untuk high-risk AI systems termasuk yang digunakan di law enforcement dan critical infrastructure. Di US beberapa kota seperti San Francisco sudah melarang penggunaan facial recognition oleh law enforcement.

Sebagai praktisi ML kita punya tanggung jawab etis untuk memastikan model yang kita develop itu fair dan tidak mendiskriminasi. Ini bukan cuma soal compliance dengan regulasi tapi juga soal dampak sosial dari teknologi yang kita buat. Kita perlu lebih aware tentang bias dalam data, melakukan fairness testing, dan transparan tentang limitasi dari sistem AI kita. Di konteks Indonesia, Mufidah et al. dalam penelitiannya menekankan bahwa implementasi AI di Indonesia perlu mempertimbangkan nilai-nilai etika lokal termasuk aspek keadilan sosial dan dampak yang tidak merata di masyarakat. Hal ini menunjukkan bahwa regulasi dan pendekatan etika AI perlu disesuaikan dengan konteks sosial-budaya masing-masing negara.

### 2.2 Dampak Energi dan Lingkungan dari Pemanfaatan AI

Saat kita ngobrol sama ChatGPT atau generate gambar pakai DALL-E jarang kepikiran berapa banyak energi dan air yang dibutuhkan di balik layar. Tapi faktanya AI punya environmental footprint yang cukup signifikan, dan ini masalah yang perlu kita address serius.

Training model AI besar membutuhkan komputasi yang massive. Patterson et al. dari Google menghitung carbon footprint dari beberapa large language models. Hasil mereka cukup membuka mata, training GPT-3 menghasilkan sekitar 502 metric tons CO₂ equivalent, setara dengan emisi dari 112 mobil bensin selama setahun. Model yang lebih besar seperti dense neural networks bisa mengkonsumsi 10 kali lebih banyak energi dibanding sparse models untuk mencapai akurasi yang sama.

Yang lebih concerning training itu cuma sekali. Setelah deployed, inference atau actual usage dari model AI mengkonsumsi energi terus-menerus. Li et al. memperkirakan bahwa data center yang run AI workloads mengkonsumsi sekitar 10-15% dari total energy usage perusahaan tech besar seperti Google, dengan 60% untuk inference dan 40% untuk training. Proyeksi ke depan lebih alarming, dengan pertumbuhan exponential dari AI usage diperkirakan data centers bisa mengkonsumsi 9% dari total electricity generation di US pada 2030. Kalau mayoritas listrik ini masih dari fossil fuels carbon footprint-nya akan sangat besar.

Di Indonesia sendiri situasinya cukup concerning. PLN mencatat konsumsi listrik dari 159 data center customers mencapai 2,378 GWh dengan hampir semua kapasitas terkonsentrasi di Jawa, Madura, dan Bali. Mengingat Indonesia masih heavily dependent pada coal-fired power plants sekitar 60% dari total electricity generation, ekspansi data center untuk AI workloads berkontribusi signifikan pada emisi karbon nasional.

Aspek yang sering overlooked adalah water consumption. Data centers butuh cooling system yang intensif untuk menjaga server tetap dingin. Li et al. dalam paper mereka Making AI Less Thirsty mengestimasi bahwa training GPT-3 mengkonsumsi sekitar 700,000 liter air bersih untuk cooling. Setiap 10-50 query ke ChatGPT membutuhkan sekitar 500 ml air. Global AI demand diproyeksikan mengkonsumsi 4.2-6.6 billion cubic meters air pada 2027, setara dengan water withdrawal dari 4-6 Denmark.

Ini problematic karena freshwater scarcity sudah jadi salah satu pressing challenges di banyak region. Data centers biasanya menggunakan evaporative cooling yang literally menguapkan air, artinya air itu ngga bisa di-recycle. Ren et al. dalam study mereka tentang environmental impact of AI servers di US menemukan bahwa deployment AI servers bisa generate water footprint 731-1,125 million m³ per tahun antara 2024-2030. Belum lagi kalau kita hitung environmental cost dari manufacturing chips dan hardware. Produksi semiconductor untuk AI accelerators seperti GPUs dan TPUs juga energy-intensive dan menggunakan rare earth materials yang extractionnya merusak lingkungan.

Melihat dampak di atas jelas kita butuh pendekatan yang lebih sustainable. Beberapa solusi yang menurut saya promising antara lain konsep Green AI yang fokus pada developing models yang efficient. Patterson et al. menunjukkan bahwa sparse models bisa achieve performance yang sama dengan kurang dari 10% energy consumption dari dense models. Kita perlu shift mindset dari bigger is better ke efficiency matters. Dalam practice kita bisa pakai techniques seperti model pruning, quantization, dan knowledge distillation untuk reduce model size tanpa sacrifice akurasi significant. Prioritize model efficiency sebagai metric bukan cuma akurasi. Share pretrained models untuk reduce redundant training.

Solusi lain adalah renewable energy untuk data centers. Beberapa tech companies mulai commit ke renewable energy. Google claim 100% renewable energy match untuk operations mereka tapi ini masih aggregate artinya bukan real-time matching. Kita perlu data centers powered by solar, wind, atau hydroelectric. Strategic placement of data centers di locations dengan renewable energy abundant. Time-shifting of non-urgent AI computations ke periods ketika renewable energy production tinggi.

Untuk water issue bisa adopt air cooling atau liquid cooling yang ngga rely on water evaporation. Pakai alternative cooling mediums. Design data centers yang leverage natural cooling di cold regions. Menurut saya perlu ada regulation yang require AI companies untuk report carbon footprint dan water usage dari AI systems. Conduct environmental impact assessments untuk large models. Set standards untuk sustainable AI development.

Sebagai praktisi ML kita juga punya peran. Kita bisa track energy consumption dari experiments kita, tools seperti CodeCarbon bisa help. Avoid unnecessary retraining atau hyperparameter search yang excessive. Choose efficient architectures ketika possible. Advocate untuk sustainable practices di organizations kita.

Jujur waktu pertama kali baca tentang ini saya agak surprised sama magnitude dari impact-nya. Kita sering fokus ke benefits dari AI seperti automation, efficiency, innovation tapi jarang mikir tentang cost-nya. Bukan berarti kita harus stop develop AI tapi kita perlu more responsible dan intentional. Tantangannya adalah mencari balance, develop AI yang beneficial untuk society tapi ngga at the expense of environmental sustainability. Dan menurut saya ini bukan trade-off yang impossible. Dengan innovation di efficient algorithms, renewable energy, dan cooling technology kita bisa have both. Tapi itu butuh awareness, effort, dan commitment dari semua stakeholders, developers, companies, policymakers, dan users.

Climate change adalah existential threat. AI bisa jadi bagian dari solution misalnya untuk climate modeling atau optimize energy grids, tapi AI sendiri juga contribute ke problem. Kita yang develop dan use AI perlu ensure bahwa technology ini ngga memperburuk crisis yang kita coba solve.
