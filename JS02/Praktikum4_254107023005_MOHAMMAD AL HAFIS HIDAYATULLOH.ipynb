Pada praktikum keempat ini saya melakukan pra pengolahan data sederhana pada data tidak terstruktur berupa citra digital `Lenna.png`, yaitu konversi ruang warna dari BGR ke RGB, pengubahan ukuran dimensi (*resizing*) menjadi 128x128 piksel, serta konversi warna ke derajat keabuan (*grayscale*).

### Langkah 0 - Persiapan

Mengimpor library `cv2` (OpenCV) dan `matplotlib.pyplot`.

```python
import cv2
import matplotlib.pyplot as plt
```

### Langkah 1 - Load Gambar

Membaca file `Lenna.png` dengan OpenCV dan mengubah ruang warnanya dari BGR (default OpenCV) menjadi RGB agar tampil dengan warna yang tepat di Matplotlib.

```python
# Membaca gambar (BGR default di OpenCV)
img = cv2.imread("Lenna.png")

# Konversi BGR -> RGB (agar warnanya tampil benar di matplotlib)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Tampilkan
plt.imshow(img_rgb)
plt.title("Gambar Asli")
plt.axis("off")
plt.show()
```

### Langkah 2 - Resize Gambar

Mengubah ukuran gambar menjadi 128x128 piksel.

```python
# Resize gambar
img_resized = cv2.resize(img_rgb, (128, 128))

plt.imshow(img_resized)
plt.title("Gambar setelah Resize (128x128)")
plt.axis("off")
plt.show()
```

### Langkah 3 - Ubah ke Grayscale

Mengubah warna citra hasil resize ke dalam bentuk derajat keabuan (*grayscale*).

```python
# Konversi ke grayscale
img_gray = cv2.cvtColor(img_resized, cv2.COLOR_BGR2GRAY)

plt.imshow(img_gray, cmap="gray")
plt.title("Gambar Grayscale")
plt.axis("off")
plt.show()
```

Dari hasil pengolahan citra di atas, terlihat bahwa setelah gambar asli (`512x512` piksel, 3 channel warna) di-*resize* menjadi `128x128` piksel, detail tepi dan tekstur bulu pada topi terlihat sedikit lebih pecah (kurang tajam) karena jumlah pikselnya berkurang drastis, tetapi komputasinya menjadi jauh lebih ringan. Selanjutnya ketika diubah ke *grayscale*, dimensi channel warnanya tereduksi dari 3 channel menjadi 1 channel intensitas keabuan (`128x128`), yang sering dipakai apabila fitur yang ingin diekstraksi lebih berfokus pada bentuk atau tekstur dibanding informasi warna.
