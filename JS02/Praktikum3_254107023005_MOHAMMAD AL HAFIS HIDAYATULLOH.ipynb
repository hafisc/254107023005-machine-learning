Pada praktikum ketiga ini saya menggunakan file `Titanic-Dataset-fixed.csv` (data Titanic hasil imputasi) untuk melakukan pemilihan fitur (*feature selection* sederhana), mengubah variabel kategorikal menjadi numerik dengan `LabelEncoder`, dan melakukan standardisasi pada kolom `Age` dengan `StandardScaler`.

### Langkah 0 & 1 - Load Data

Mengimpor library yang dibutuhkan (`pandas`, `LabelEncoder`, `StandardScaler`) dan memuat data `Titanic-Dataset-fixed.csv`.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

dpath = 'Titanic-Dataset-fixed.csv'
df = pd.read_csv(dpath)
df.head()
```

### Langkah 2 - Slice Data

Memilih variabel yang digunakan sebagai fitur yaitu `Survived`, `Pclass`, `Age`, `Sex`, dan `Cabin`.

```python
df = df[['Survived', 'Pclass', 'Age', 'Sex', 'Cabin']]
df.head()
```

### Langkah 3 - Encoding

Melakukan *Label Encoding* pada kolom kategorikal `Sex` dan `Cabin` agar nilainya berubah menjadi representasi angka bulat (integer).

```python
le = LabelEncoder() # membuat objek dari LabelEncoder
df['Sex'] = le.fit_transform(df['Sex']) # proses encoding
df['Cabin'] = le.fit_transform(df['Cabin']) # proses encoding
```

### Langkah 4 - Verifikasi Hasil Encoding

Mengecek kembali 5 data teratas setelah proses encoding.

```python
df.head()
```

Dari hasil encoding di atas, kolom `Sex` yang tadinya berisi string `'male'` dan `'female'` sudah berubah menjadi angka `1` (`male`) dan `0` (`female`). Begitu juga dengan kolom `Cabin`, nilai `'DECK'` dikodekan menjadi `115`, `'C85'` menjadi `81`, `'C123'` menjadi `55`, dan seterusnya sesuai urutan kategori uniknya.

### Langkah 5 - Standarisasi

Melakukan standardisasi pada kolom `Age` menggunakan `StandardScaler` sehingga distribusinya memiliki rata-rata ($\mu$) mendekati 0 dan simpangan baku ($\sigma$) mendekati 1.

```python
std = StandardScaler()
df['Age'] = std.fit_transform(df[['Age']])
```

### Langkah 6 - Verifikasi Hasil Standarisasi

Mengecek kembali 5 data teratas setelah proses standardisasi.

```python
df.head()
```

Setelah dilakukan standardisasi dengan `StandardScaler`, nilai pada kolom `Age` kini berada di sekitar angka 0 (ada yang bernilai negatif untuk penumpang yang usianya di bawah rata-rata 29.70 tahun seperti umur 22 tahun yang menjadi `-0.592481` dan umur 26 tahun menjadi `-0.284663`, serta bernilai positif untuk penumpang di atas rata-rata seperti umur 38 tahun menjadi `0.638789` dan umur 35 tahun menjadi `0.407926`).
