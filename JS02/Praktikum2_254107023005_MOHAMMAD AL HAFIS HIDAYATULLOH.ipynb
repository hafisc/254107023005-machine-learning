Pada praktikum kedua ini saya melakukan penanganan nilai yang hilang (*data imputation*) pada dataset Titanic, khususnya pada kolom `Age`, `Cabin`, dan `Embarked` yang pada Praktikum 1 terdeteksi memiliki nilai `NaN`.

### Langkah 0 - Persiapan

Mengimpor library `pandas` dan `numpy`.

```python
import pandas as pd
import numpy as np
```

### Langkah 1 - Memuat Data

Membaca dataset `Titanic-Dataset.csv` dan menampilkan 5 data teratas.

```python
df = pd.read_csv('Titanic-Dataset.csv')
# Tampilkan 5 data teratas
df.head()
```

### Langkah 2 - Inspeksi Data

Mengecek informasi kolom dan jumlah data yang hilang sebelum dilakukan proses imputasi.

```python
df.info()
```

```python
df.isnull().sum()
```

### Langkah 3 - Data Imputation

Mengisi nilai kosong dengan strategi berikut:
- `Age` diisi menggunakan nilai rata-rata (*mean*) yaitu sekitar `29.699118`.
- `Cabin` diisi dengan string `"DECK"` dengan asumsi penumpang tanpa nomor kabin berada di dek kapal.
- `Embarked` diisi menggunakan nilai modus (*mode*) yaitu `'S'` (Southampton).

*Catatan: Pada Pandas versi terbaru (Copy-on-Write), penugasan `df['col'] = df['col'].fillna(...)` digunakan supaya hasil pengisian langsung tersimpan pada DataFrame utama.*

```python
# Age - mean
df['Age'] = df['Age'].fillna(value=df['Age'].mean())

# Cabin - "DECK"
df['Cabin'] = df['Cabin'].fillna(value="DECK")

# Embarked - modus
df['Embarked'] = df['Embarked'].fillna(value=df['Embarked'].mode()[0])
```

### Langkah 4 - Validasi Hasil

Mengecek kembali dengan `df.info()`, `df.isnull().sum()`, dan menampilkan 10 baris pertama untuk memastikan seluruh nilai `NaN` sudah terisi.

```python
df.info()
```

```python
df.isnull().sum()
```
